```python
from rdkit import Chem
import sascorer  # Ensure this is available in your working directory or Python path
```

```python
def get_sa_score(smiles: str) -> float:
    """
    Compute the Synthetic Accessibility (SA) score for a given SMILES string.
    
    Parameters:
        smiles (str): A valid SMILES string.
    
    Returns:
        float: SA score (lower is easier to synthesize). Returns None if invalid SMILES.
    """
    mol = Chem.MolFromSmiles(smiles)
    if mol is None:
        print(f"Invalid SMILES: {smiles}")
        return None
    return sascorer.calculateScore(mol)
```
